```python
import numpy as np
import pandas as pd, os, datetime
import statsmodels.api as sm
import statsmodels.formula.api as smf

# Statistical analysis
from multiprocessing import Pool
from scipy import stats
import math

# Import the loader function
os.chdir('/g/data/ng72/ms5578/ID_HW_BARRA')
from process_code import load_generation_data
```

```python
clust_map = pd.read_csv("data/preprocess/wind_spatial_clusters.csv")
```

```python
data, info = load_generation_data(
    sdate="2009-07-01",
    edate="2024-06-30",
    mode="hourly",
    ftype=["Wind"],
    apply_remove_negatives=True,
    apply_remove_wind_zeros=True,
    apply_min_heatwave_days=True,
    min_heatwave_days_threshold=20,
    apply_clear_agc=False
)
```

```python
df = pd.merge(data,clust_map[['DUID','cluster']], on='DUID').sort_values(['EHF_flag','cluster','DUID','time'])
```

```python
def profile_matrix(group_df):
    # Add day and hour columns
    group_df = group_df.copy()
    group_df['day'] = group_df['time'].dt.floor('D')
    group_df['hour'] = group_df['time'].dt.hour

    # Pivot: rows=days, cols=hours
    daily = group_df.pivot_table(
        index='day',
        columns='hour',
        values='TOTALMWh'
    ).sort_index()

    return daily.dropna(axis=0).to_numpy()
```

```python
def bootstrap_curve(X, n_boot=500):
    """
    Bootstrap mean and median daily curves with confidence intervals.
    
    Parameters
    ----------
    X : np.ndarray
        Array of shape (n_days, n_hours)
    n_boot : int
        Number of bootstrap samples

    Returns
    -------
    mean_curve : np.ndarray, shape (n_hours,)
    mean_lower : np.ndarray, shape (n_hours,)
    mean_upper : np.ndarray, shape (n_hours,)
    median_curve : np.ndarray, shape (n_hours,)
    median_lower : np.ndarray, shape (n_hours,)
    median_upper : np.ndarray, shape (n_hours,)
    """
    n_days, n_hours = X.shape
    curves_mean = np.empty((n_boot, n_hours))
    curves_median = np.empty((n_boot, n_hours))
    rng = np.random.default_rng(42)

    for b in range(n_boot):
        idx = rng.integers(0, n_days, size=n_days)
        sample = X[idx]
        curves_mean[b, :] = sample.mean(axis=0)
        curves_median[b, :] = np.median(sample, axis=0)

    # Compute mean and 95% CI
    mean_curve = curves_mean.mean(axis=0)
    mean_lower = np.percentile(curves_mean, 2.5, axis=0)
    mean_upper = np.percentile(curves_mean, 97.5, axis=0)

    # Compute median and 95% CI
    median_curve = curves_median.mean(axis=0)
    median_lower = np.percentile(curves_median, 2.5, axis=0)
    median_upper = np.percentile(curves_median, 97.5, axis=0)

    return mean_curve, mean_lower, mean_upper, median_curve, median_lower, median_upper
```

```python
def process_group(group_tuple):
    name, gdf = group_tuple
    gdf = gdf.copy()
    gdf['day'] = gdf['time'].dt.floor('D')
    gdf['hour'] = gdf['time'].dt.hour
    daily_matrix = gdf.pivot_table(index='day', columns='hour', values='TOTALMWh').dropna().sort_index()
    X = daily_matrix.to_numpy()
    
    return name, bootstrap_curve(X, n_boot=10000)

# Prepare groups
groups = list(df.groupby(['cluster', 'EHF_flag']))

# 6 clusters, 2 modes = 12 processes
with Pool(processes=12) as pool:
    results = pool.map(process_group, groups)
```

```python
rows = []
for (cluster, hw_status), curves in results:
    mean_curve, mean_lower, mean_upper, median_curve, median_lower, median_upper = curves
    for hour in range(len(mean_curve)):
        rows.append({
            'cluster': cluster,
            'hw_status': hw_status,
            'hour': hour,
            'mean': mean_curve[hour],
            'mean_lower': mean_lower[hour],
            'mean_upper': mean_upper[hour],
            'median': median_curve[hour],
            'median_lower': median_lower[hour],
            'median_upper': median_upper[hour]
        })

df_results = pd.DataFrame(rows)
```

```python
import matplotlib.pyplot as plt

def plot_cluster_with_ci(df_results, cluster_name, metric='mean'):
    """
    Plot daily curves with confidence intervals for all hw_status in a cluster.
    
    Parameters
    ----------
    df_results : pd.DataFrame
        Tidy dataframe with columns:
        ['cluster', 'hw_status', 'hour', 'mean', 'mean_lower', 'mean_upper',
         'median', 'median_lower', 'median_upper']
    cluster_name : int or str
        The cluster to plot
    metric : str
        Either 'mean' or 'median' (which curve to plot)
    """
    if metric not in ['mean', 'median']:
        raise ValueError("metric must be 'mean' or 'median'")

    cluster_df = df_results[df_results['cluster'] == cluster_name]
    hw_statuses = cluster_df['hw_status'].unique()
    
    plt.figure(figsize=(10,5))
    colors = ['blue', 'orange', 'green', 'red', 'purple', 'brown']  # extend if needed

    for i, status in enumerate(hw_statuses):
        status_df = cluster_df[cluster_df['hw_status'] == status].sort_values('hour')

        # Select the columns based on metric
        y = status_df[metric].values
        lower = status_df[f'{metric}_lower'].values
        upper = status_df[f'{metric}_upper'].values

        plt.plot(status_df['hour'], y, color=colors[i % len(colors)],
                 label=f'{metric.capitalize()} ({status})')
        plt.fill_between(status_df['hour'], lower, upper, color=colors[i % len(colors)], alpha=0.3)

    plt.xlabel('Hour of Day')
    plt.ylabel('Value')
    plt.title(f'{metric.capitalize()} daily curve with CI for cluster {cluster_name}')
    plt.xticks(range(0,24))
    plt.legend()
    plt.show()
    
plot_cluster_with_ci(df_results, cluster_name=5, metric='median')
```
