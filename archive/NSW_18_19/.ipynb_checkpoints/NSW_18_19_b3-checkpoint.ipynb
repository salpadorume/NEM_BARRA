```python
def jitter(group):
    duplicates = group.groupby(['lat', 'lon']).cumcount()
    
    # Jitter function: add a small random offset
    np.random.seed(42)  # For reproducibility
    jitter_strength = 0.05 # Adjust as needed; degrees latitude/longitude
    
    group['lat_jittered'] = group['lat'] + np.random.uniform(-jitter_strength, jitter_strength, size=len(group)) * duplicates
    group['lon_jittered'] = group['lon'] + np.random.uniform(-jitter_strength, jitter_strength, size=len(group)) * duplicates

    return group
```

```python
# A temporary patch to fix fuel types
gen_details['fuel_source_primary'] = gen_details['fuel_source_primary'].replace({'Solar - Solar': 'Solar','Wind - Wind': 'Wind'})
gen_details = jitter(gen_details)
```

```python
hw_tseries = pd.read_csv(f"{ehf_fpath}/gen_hw_status.csv")
```

```python
def process_group(grp, gen_fpath, hw_tseries, start_date=None, end_date=None):
    """
    Processes a single group of generator data.

    Parameters:
        grp (pd.DataFrame): A single group from gen_details (e.g., from groupby('region')).
        gen_fpath (str): Path to directory containing CSV files named by DUID (e.g., DUID.csv).
        hw_tseries (pd.DataFrame): DataFrame with columns 'time', 'DUID', and data to merge.
        start_date (str): Start date for subsetting the time series.
        end_date (str): End date for subsetting the time series.

    Returns:
        pd.DataFrame: The merged result for the group.
    """
    safe_duids = [duid.replace("/", "_").replace("\\", "_") for duid in grp['DUID']]
    gen_locs = [f"{gen_fpath}/{duid}.csv" for duid in safe_duids]
    dfs = [pd.read_csv(fp,dtype='object') for fp in gen_locs if os.path.exists(fp)]
    print(f"Loaded {len(dfs)} CSV file(s) out of {len(gen_locs)} expected.")

    if not dfs:
        return None

    # This is in case of accidental mid-file headers
    dfs = pd.concat(dfs, ignore_index=True)
    header_row = dfs.columns.tolist()
    dfs = dfs[~dfs.apply(lambda row: list(row) == header_row, axis=1)]

    # This is to correct the column types after removing header rows
    dfs['time'] = pd.to_datetime(dfs['time'])
    dfs['TOTALCLEARED'] = dfs['TOTALCLEARED'].astype(float)
    dfs['TOTALMWh'] = dfs['TOTALMWh'].astype(float)
    dfs['AGCSTATUS'] = pd.to_numeric(dfs['AGCSTATUS'], errors='coerce').fillna(0).astype(int)
    
    dfs['time'] = pd.to_datetime(dfs['time'])
    dfs = dfs.set_index('time').sort_index()
    dfs = dfs.loc[start_date:end_date]

    # Group by DUID and hourly time, sum TOTALMWh
    agg_func = {'TOTALMWh':'sum','TOTALCLEARED':'sum','AGCSTATUS':'max'}
    grouped = dfs.groupby(['DUID', pd.Grouper(freq='1d')]).agg(agg_func).reset_index()

    hw_tseries['time'] = pd.to_datetime(hw_tseries['time'])
    hw_tseries = hw_tseries.set_index(['time']).sort_index()
    hw_tseries = hw_tseries.loc[sdate:edate]
    hw_tseries = hw_tseries.reset_index().set_index(['DUID','time']).sort_index()

    merged = pd.merge_asof(
        grouped.sort_values(by=['time', 'DUID']),
        hw_tseries.sort_values(by=['time', 'DUID']),
        by='DUID',
        on='time',
        tolerance=pd.Timedelta("1d"),
        direction='nearest'
    )

    merged = merged.dropna(how='all')

    return merged
```

```python
def select_group(gen_details=gen_details, state=None,ftype=None):
    if state is not None and ftype is not None:
        groups = gen_details.groupby(['region','fuel_source_primary'])
        grp = groups.get_group((state,ftype))
    elif state is not None:
        groups = gen_details.groupby('region')
        grp = groups.get_group(state)
    elif ftype is not None:
        groups = gen_details.groupby('fuel_source_primary')
        grp = groups.get_group(ftype)

    else:
        grp = gen_details

    return grp
```

```python
demand = pd.read_csv('/scratch/ng72/ms5578/time_series/state_demand.csv')
demand['time'] = pd.to_datetime(demand['time'])
demand = demand.set_index('time').sort_index()
demand = demand.loc[sdate:edate]
demand = demand.loc[demand['REGIONID'].isin(['NSW1','QLD1'])]
demand = demand.groupby(['REGIONID']).resample('D', level=0).agg({'TOTALDEMAND':'sum'}).reset_index()
```

```python
def clean_df(df, gen_details):
    df = df.merge(gen_details[['DUID', 'reg_cap_mw','technology_type_primary','fuel_source_primary','lat_jittered','lon_jittered']], on='DUID', how='left')
    df = df[~((df['TOTALMWh'] < 0))]
    mask = (df['fuel_source_primary'].isin([
            'Water', 'Natural Gas Pipeline', 'Black Coal', 'Coal Seam Methane',
            'Brown Coal', 'Diesel', 'Kerosene'
        ]) & (df['TOTALCLEARED'] <= 0))
    df.loc[mask, 'TOTALMWh'] = np.nan
    df['reg_cap_mw'] = df['reg_cap_mw'].astype('float') * 24
    df['cap_norm'] = df['TOTALMWh']/df['reg_cap_mw']
    df['min_max'] = df.groupby('DUID')['TOTALMWh'].transform(lambda x: (x - x.min()) / (x.max() - x.min()))
    return df
```

```python
def plot_day_count(df, group, cmap, cleared_only=True):
    # Same filtering + processing as before
    if cleared_only == True:
        df = df[~((df['fuel_source_primary'].isin([
            'Water', 'Natural Gas Pipeline', 'Black Coal', 'Coal Seam Methane',
            'Brown Coal', 'Diesel', 'Kerosene'])) & (df['TOTALCLEARED'] <= 0))]

    day_counts = df.groupby(['DUID', 'EHF_flag']).size().unstack(fill_value=0).reset_index()
    day_counts.columns.name = None
    day_counts = day_counts.rename(columns={0: 'no_hw', 1: 'in_hw'})

    day_counts = day_counts.merge(group[['DUID', 'fuel_source_primary', 'station_name']], on='DUID', how='left')
    day_counts = day_counts.sort_values(by=['fuel_source_primary', 'DUID'])
    day_counts['color'] = day_counts['fuel_source_primary'].map(cmap)

    fig = go.Figure()

    # Bottom bars: fuel color for "Not in Heatwave"
    fig.add_trace(go.Bar(
        x=day_counts['DUID'],
        y=day_counts['no_hw'],
        name='Not in Heatwave',
        marker_color=day_counts['color'],
        customdata=day_counts[['fuel_source_primary', 'station_name']].values,
        hovertemplate=(
            '<b>DUID:</b> %{x}<br>' +
            '<b>Name:</b> %{customdata[1]}<br>' +
            '<b>Days (Not in HW):</b> %{y}<br>' +
            '<b>Fuel:</b> %{customdata[0]}<extra></extra>'
        )
    ))

    # Top bars: grey for "In Heatwave"
    fig.add_trace(go.Bar(
        x=day_counts['DUID'],
        y=day_counts['in_hw'],
        name='In Heatwave',
        marker_color='grey',
        customdata=day_counts[['fuel_source_primary', 'station_name']].values,
        hovertemplate=(
            '<b>DUID:</b> %{x}<br>' +
            '<b>Name:</b> %{customdata[1]}<br>' +
            '<b>Days (In HW):</b> %{y}<br>' +
            '<b>Fuel:</b> %{customdata[0]}<extra></extra>'
        )
    ))

    fig.update_layout(
        title='Heatwave vs Non-Heatwave Days by Generator',
        xaxis_title='Generator (DUID)',
        yaxis_title='Number of Days',
        barmode='stack',
        height=600,
        xaxis=dict(
            tickmode='array',
            tickvals=day_counts['DUID'],
            ticktext=day_counts['DUID'],
            tickangle=90
        ),
        legend=dict(
            title='Status',
            traceorder='normal'
        )
    )

    fig.show()
    return day_counts
```

```python
def plot_hw_frac(df):
    # --- 1. Group data ---
    # % of generators in/out of heatwave
    gens_in_hw = df.groupby('time')['EHF_flag'].value_counts().unstack(fill_value=0)
    gens_percent = gens_in_hw.div(gens_in_hw.sum(axis=1), axis=0) * 100
    
    # Total generation split by EHF_flag
    gen_split = df.groupby(['time', 'EHF_flag'])['TOTALMWh'].sum().unstack(fill_value=0)
    
    
    # --- Heatwave Participation (% stacked bars) ---
    fig = go.Figure()
    
    fig.add_trace(go.Scatter(
        x=gens_percent.index.astype(str),
        y=gens_percent[0],
        mode='lines',
        name='Not in Heatwave (%)',
        line=dict(color='lightblue'),
        stackgroup='one'
    ))
    
    fig.add_trace(go.Scatter(
        x=gens_percent.index.astype(str),
        y=gens_percent[1],
        mode='lines',
        name='In Heatwave (%)',
        line=dict(color='tomato'),
        stackgroup='one'
    ))
    
    fig.update_layout(
        title='Percentage of Generators in Heatwave per Day',
        xaxis_title='Date',
        yaxis_title='Participation (%)',
        yaxis=dict(ticksuffix='%', range=[0, 100]),
        height=400,
        legend=dict(title='EHF_flag'),
        hovermode='x unified'
    )
    
    fig.show()
```

```python
def plot_hw_frac_multi(dfs, labels, colors=None):
    """
    Plot percentage of generators IN heatwave (EHF_flag == 1) for multiple states, with shaded areas.
    
    Parameters:
    - dfs: list of pandas DataFrames (must contain 'time', 'EHF_flag')
    - labels: list of state/region names
    - colors: list of line colors for each state (optional)
    """
    fig = go.Figure()

    if colors is None:
        colors = ['tomato', 'royalblue', 'mediumseagreen']

    for df, label, color in zip(dfs, labels, colors):
        # Group by time and count EHF_flag values
        gens_in_hw = df.groupby('time')['EHF_flag'].value_counts().unstack(fill_value=0)
        gens_percent = gens_in_hw.div(gens_in_hw.sum(axis=1), axis=0) * 100
        
        # Check if there's any EHF_flag == 1 data
        if 1 in gens_percent.columns:
            fig.add_trace(go.Scatter(
                x=gens_percent.index.astype(str),
                y=gens_percent[1],
                mode='lines',
                name=f"{label} - In Heatwave",
                line=dict(color=color),
                fill='tozeroy',  # <--- this shades under the line
                fillcolor=color.replace(')', ', 0.3)').replace('rgb', 'rgba') if 'rgb' in color else None
            ))

    fig.update_layout(
        title='Percentage of Generators in Heatwave per Day by State',
        xaxis_title='Date',
        yaxis_title='In Heatwave (%)',
        yaxis=dict(ticksuffix='%', range=[0, 100]),
        height=500,
        hovermode='x unified',
        legend=dict(title='State')
    )

    fig.show()
```

```python
plot_hw_frac_multi(
    dfs=[vic_df, nsw_df, qld_df],
    labels=['Victoria', 'New South Wales', 'Queensland']
)
```

```python
def plot_violin(df):
    df = df[~((df['fuel_source_primary'].isin([
            'Water', 'Natural Gas Pipeline', 'Black Coal', 'Coal Seam Methane',
            'Brown Coal', 'Diesel', 'Kerosene'])) & (df['TOTALCLEARED'] <= 0))].copy()
    df = df[~(df['fuel_source_primary'] == 'Kerosene')].copy()
    df['in_hw'] = df['EHF_flag'].map({1: 'Heatwave', 0: 'Non-Heatwave'})
    df = df.sort_values('EHF_flag')
    
    # Step 3: Create violin plot grouped by fuel type and split by heatwave status
    fig = px.violin(
        df,
        x='fuel_source_primary',
        y='min_max',
        color='in_hw',
        box=True,
        points='all',
        hover_data=['DUID', 'time','TOTALMWh'],
        labels={
            'fuel_source_primary': 'Fuel Source',
            'gen_pct_of_cap': 'Generation (% of Registered Capacity)',
            'in_hw': 'Condition',
            'TOTALMWh':'Daily Generation (MWh)'
        },
        title='Daily Normalised Generation: Heatwave vs Non-Heatwave by Fuel Source',
    )
    
    fig.update_layout(
        yaxis_title='Min-max Normalised Daily Generation',
        xaxis_title='Fuel Source',
        height=600,
        legend_title='Condition'
    )
    
    fig.show()
```

```python
def hist_dropdown_state(df1,df2,keys=['NSW1','QLD1']):
    
    df = pd.concat([df1, df2], keys=keys)
    df = df.reset_index(level=0, names='region')
    
    df = df[~((df['fuel_source_primary'].isin([
            'Water', 'Natural Gas Pipeline', 'Black Coal', 'Coal Seam Methane',
            'Brown Coal', 'Diesel', 'Kerosene'])) & (df['TOTALCLEARED'] <= 0))].copy()
    df = df[~(df['fuel_source_primary'].isin(['Kerosene','Coal Seam Methane']))].copy()
    
    fuel_types = df['fuel_source_primary'].unique()
    
    fig = go.Figure()
    trace_labels = []
    
    # Step 1: Add histogram and KDE curve per fuel type
    for fuel in fuel_types:
        for label, flag in [('region', keys[0]), ('region', keys[1])]:
            color = 'green' if flag =='QLD1' else 'blue'
            data = df[(df['fuel_source_primary'] == fuel) & (df['region'] == flag)]['TOTALMWh'].dropna()
    
            # Histogram
            fig.add_trace(
                go.Histogram(
                    x=data,
                    name=f'{fuel} - {flag}',
                    opacity=0.6,
                    visible=True,  # default to visible for "All"
                    histnorm='probability density',
                    nbinsx=50,
                    xbins=dict(
                            start=0,        
                            end=data.max(),    
                            size=(data.max() - 0) / 50),
                    marker_color=color
                )
            )
            trace_labels.append((fuel, flag))
    
            # KDE curve
            if len(data) > 1:
                kde = gaussian_kde(data)
                x_vals = np.linspace(data.min(), data.max(), 200)
                y_vals = kde(x_vals)
    
                fig.add_trace(
                    go.Scatter(
                        x=x_vals,
                        y=y_vals,
                        mode='lines',
                        name=f'{fuel} - {flag} KDE',
                        visible=True,
                        line=dict(color=color, dash='dash')
                    )
                )
                trace_labels.append((fuel, f'{flag}_KDE'))
    
    # Step 2: Create dropdown buttons
    dropdown_buttons = []
    
    # All: show everything
    dropdown_buttons.append(
        dict(
            label='All',
            method='update',
            args=[
                {"visible": [True] * len(trace_labels)},
                {"title": {"text": "Unit-Normalized Daily Energy Generation – All Fuel Types"}}
            ]
        )
    )
    
    # One button per fuel type
    for fuel in fuel_types:
        visible = [(f == fuel) for f, _ in trace_labels]
        dropdown_buttons.append(
            dict(
                label=fuel,
                method='update',
                args=[
                    {"visible": visible},
                    {"title": {"text": f"Daily Energy Generation of NSW and QLD – {fuel} (HW vs non-HW + KDEs)"}}
                ]
            )
        )
    
    # Step 3: Layout
    fig.update_layout(
        title="Daily Energy Generation of NSW and QLD",
        xaxis_title="Generation (MWh)",
        yaxis_title="Density",
        
        barmode="overlay",
        updatemenus=[
            dict(
                buttons=dropdown_buttons,
                direction='down',
                x=1.05,
                xanchor='left',
                y=1.15,
                yanchor='top'
            )
        ]
    )
    
    fig.show()
```

```python
def line_total_gen(df1, df2, region1_name='Region 1', region2_name='Region 2'):
    # Label and merge dataframes
    df1 = df1.copy()
    df2 = df2.copy()
    df1['region'] = region1_name
    df2['region'] = region2_name
    df_all = pd.concat([df1, df2], ignore_index=True)

    # Compute heatwave % per region
    hw_percent_all = []
    for region, df_region in df_all.groupby('region'):
        total_by_day = df_region.groupby('time')['DUID'].nunique()
        hw_by_day = df_region[df_region['EHF_flag'] == 1].groupby('time')['DUID'].nunique()
        hw_percent = (hw_by_day / total_by_day * 100).fillna(0).reset_index(name='hw_percent')
        hw_percent['region'] = region
        hw_percent_all.append(hw_percent)
    hw_percent_df = pd.concat(hw_percent_all)

    # Aggregate energy generation
    df_energy = df_all.groupby(['time', 'fuel_source_primary', 'region'])['TOTALMWh'].sum().reset_index()

    # Create base figure
    fig = go.Figure()

    # Add energy generation lines
    for (region, fuel), group in df_energy.groupby(['region', 'fuel_source_primary']):
        fig.add_trace(go.Scatter(
            x=group['time'],
            y=group['TOTALMWh'],
            mode='lines',
            name=f'{region} - {fuel}',
            yaxis='y1'
        ))

    # Add heatwave shading (secondary y-axis)
    for region in hw_percent_df['region'].unique():
        region_data = hw_percent_df[hw_percent_df['region'] == region]
        fig.add_trace(go.Scatter(
            x=region_data['time'],
            y=region_data['hw_percent'],
            mode='lines',
            fill='tozeroy',
            name=f'{region} % in Heatwave',
            yaxis='y2',
            line=dict(width=0),
            fillcolor='rgba(255,100,100,0.25)',
            hoverinfo='x+y'
        ))

    # Max heatwave % for axis scaling
    max_hw = hw_percent_df['hw_percent'].max()

    # Layout with dual axes
    fig.update_layout(
        title='Daily Energy Generation and Heatwave % by Region',
        xaxis=dict(title='Date'),
        yaxis=dict(
            title='Energy Generation (MWh)',
            side='left'
        ),
        yaxis2=dict(
            title='% in Heatwave',
            overlaying='y',
            side='right',
            showgrid=False,
            ticksuffix='%',
            range=[0, max(5, max_hw * 1.2)]
        ),
        legend=dict(
            x=1.05,
            y=1,
            xanchor='left',
            yanchor='top'
        ),
        height=600
    )

    fig.show()
```

```python
def market_share(df, name,cmap):
    # Step 0: Compute heatwave percentage per day
    total_by_day = df.groupby('time')['DUID'].nunique()
    hw_by_day = df[df['EHF_flag'] == 1].groupby('time')['DUID'].nunique()
    hw_percent = (hw_by_day / total_by_day * 100).fillna(0).reset_index(name='hw_percent')
    
    # Step 1: Aggregate generation by time and fuel
    gen_by_fuel = df.groupby(['time', 'fuel_source_primary'])['TOTALMWh'].sum().reset_index()
    
    # Step 2: Calculate total generation per time
    total_gen = df.groupby('time')['TOTALMWh'].sum().reset_index().rename(columns={'TOTALMWh': 'total_generation'})
    
    # Step 3: Merge to compute % share
    gen_share = pd.merge(gen_by_fuel, total_gen, on='time')
    gen_share['percent_share'] = 100 * gen_share['TOTALMWh'] / gen_share['total_generation']
    gen_share['color']  = gen_share['fuel_source_primary'].map(cmap)
    
    # Step 4: Plot as area or line chart
    fig = px.area(
        gen_share,
        x='time',
        y='percent_share',
        color='fuel_source_primary',  # ✅ group and label by fuel
        color_discrete_map=cmap,      # ✅ map to your custom colors
        labels={
            'time': 'Date',
            'percent_share': 'Market Share (%)',
            'fuel_source_primary': 'Fuel Source'
        },
        title=f'{name} Daily Generation Market Share by Fuel Source (%)'
    )
    
    fig.update_layout(
        yaxis_title='Market Share (%)',
        xaxis_title='Date',
        height=600,
        legend_title='Fuel Source'
    )
    
    fig.add_trace(go.Scatter(
        x=hw_percent['time'],
        y=hw_percent['hw_percent'],
        mode='lines',
        line=dict(color='red', dash='dash'),
        name='% in Heatwave',
        yaxis='y2',
        hoverinfo='x+y'
    ))
    
    # --- Step 4: Configure secondary y-axis and layout ---
    fig.update_layout(
        yaxis=dict(
            title='Market Share (%)',
            range=[0, 100]
        ),
        yaxis2=dict(
            title='% in Heatwave',
            overlaying='y',
            side='right',
            range=[0, 100],
            showgrid=False
        ),
        legend=dict(
            x=1.1,           # Position to the right of the plot
            y=1,
            xanchor='left',
            yanchor='top'
        ),
        xaxis_title='Date',
        height=600,
        legend_title='Fuel Source'
        
    )
    
    
    
    fig.show()
```

```python
market_share(nsw_df,'NSW',cmap)
```

```python
market_share(qld_df, 'QLD',cmap)
```

```python
def plot_loc(df, info, cmap):
    unit_summ = df.reset_index().groupby(['DUID','EHF_flag'])['min_max'].agg(['mean','median','min','max','std','var']).reset_index()
    unit_summ = unit_summ.merge(info, on='DUID',how='left')
    unit_summ['size'] = unit_summ['mean'].abs()
    unit_summ['size'] = unit_summ['size'].replace(0, 0.001)  # replace zeros with small positive value
    unit_summ['size'] = unit_summ['size']*100  # replace zeros with small positive value
    
    in_heatwave = unit_summ[unit_summ['EHF_flag'] == True]
    no_heatwave = unit_summ[unit_summ['EHF_flag'] == False]
    
    # Create traces using px.scatter_map
    fig = px.scatter_map(
        in_heatwave,
        lat='lat_jittered',
        lon='lon_jittered',
        hover_name='DUID',
        hover_data={
            'lat': False,
            'lon': False,
            'lat_jittered': False,
            'lon_jittered': False,
            'fuel_source_primary': True,
            'size':False
        },
        color='fuel_source_primary',
        color_discrete_map=cmap,
        zoom=5,
        map_style='carto-darkmatter'
    )
    
    fig.update_layout(
        legend=dict(
            title="Fuel Type",
        ),
        title_text="Locations of Generation Units"
    )
    
    fig.show()
```

```python
def animap(df):
    # Drop rows with missing size values
    df.dropna(subset=['min_max'], inplace=True)

    # Duplicate entries for halo effect
    df_halo = df[df['EHF_flag'] == 1].copy()
    df_halo['min_max'] *= 2             # make halo larger
    df_halo['opacity'] = 0.2            # transparent halo
    df_halo['layer'] = 'halo'
    df_halo['marker_color'] = 'lightblue'  # set halo color

    # Main markers
    df_main = df.copy()
    df_main['opacity'] = 1.0
    df_main['layer'] = 'main'
    df_main['marker_color'] = df_main['fuel_source_primary']  # category-based coloring

    # Combine layers
    df_all = pd.concat([df_halo, df_main], ignore_index=True)

    # Plot
    fig = px.scatter_geo(
        df_all,
        lat='lat_jittered',
        lon='lon_jittered',
        animation_frame='time',
        size='min_max',
        color='marker_color',  # use the custom color column
        opacity=0.6,
        projection='natural earth',
        scope='world',
        hover_name='DUID'
    )

    # Remove marker border lines
    fig.update_traces(marker=dict(line=dict(width=0)))

    # Zoom in on Australia
    fig.update_geos(
        lataxis_range=[-45, -10],
        lonaxis_range=[110, 155],
        projection_type='natural earth',
        showland=True, landcolor='rgb(240,240,240)',
        showcountries=True, countrycolor="Black"
    )

    fig.update_layout(
        title='Animated Geo Plot with Light Blue Halo Highlights',
        geo=dict(
            resolution=50,
            showland=True,
            landcolor="rgb(229, 229, 229)",
            showcountries=True,
            countrycolor="Black"
        )
    )

    fig.show()
```

```python
animap(qld_df)
```

```python
def plot_gen_over_time(df):
    """
    Plot generation over time during and outside of heatwaves,
    with a dropdown to filter by fuel source.
    """
    # Step 1: Aggregate total generation by time, fuel type, and heatwave flag
    gen = df.groupby(['time', 'fuel_source_primary', 'EHF_flag'])['min_max'].mean().reset_index()

    # Step 2: Pivot to get two columns: heatwave and non-heatwave per time
    pivot = gen.pivot_table(index=['time', 'fuel_source_primary'],
                            columns='EHF_flag',
                            values='min_max',
                            fill_value=0).reset_index()
    pivot.columns = ['time', 'fuel_source_primary', 'non_hw_gen', 'hw_gen']  # assumes EHF_flag 0 then 1

    # Step 3: Prepare figure and dropdown options
    fuel_types = pivot['fuel_source_primary'].dropna().unique()
    fig = go.Figure()

    for i, fuel in enumerate(fuel_types):
        data = pivot[pivot['fuel_source_primary'] == fuel].sort_values('time')

        fig.add_trace(go.Scatter(
            x=data['time'],
            y=data['non_hw_gen'],
            mode='lines',
            name='Non-Heatwave',
            visible=True if i == 0 else False
        ))

        fig.add_trace(go.Scatter(
            x=data['time'],
            y=data['hw_gen'],
            mode='lines',
            name='Heatwave',
            visible=True if i == 0 else False
        ))

    # Step 4: Create dropdown buttons
    buttons = []
    for i, fuel in enumerate(fuel_types):
        visibility = [False] * (2 * len(fuel_types))
        visibility[2 * i] = True      # Non-heatwave line
        visibility[2 * i + 1] = True  # Heatwave line

        buttons.append(dict(
            label=fuel,
            method='update',
            args=[{'visible': visibility},
                  {'title': f'Generation Over Time — {fuel}'}]
        ))

    # Step 5: Layout
    fig.update_layout(
        updatemenus=[dict(
            active=0,
            buttons=buttons,
            x=1.05,
            xanchor='left',
            y=1.1,
            yanchor='top'
        )],
        title='Generation Over Time — by Fuel Type',
        xaxis_title='Time',
        yaxis_title='Total Generation (MWh)',
        height=600
    )

    fig.show()
```

```python
plot_gen_over_time(qld_df)
```

```python
def barplot_gen_per_unit(df, demand_df, both_days=False):
    # Step 1: Aggregate to get total gen per DUID per day
    df_day_unit = df.groupby(['time', 'fuel_source_primary', 'EHF_flag', 'DUID'])['min_max'].sum().reset_index()

    # Step 2: Compute mean per-unit generation by day, fuel, and heatwave flag
    df_mean = df_day_unit.groupby(['time', 'fuel_source_primary', 'EHF_flag'])['min_max'].mean().reset_index()

    # Step 3: Keep only time + fuel groups that have both HW and non-HW
    valid_groups = (
        df_mean.groupby(['time', 'fuel_source_primary'])['EHF_flag']
        .nunique()
        .reset_index(name='hw_presence')
    )
    valid_groups = valid_groups[valid_groups['hw_presence'] == 2][['time', 'fuel_source_primary']]


    if both_days == True:
        # Filter df_mean to keep only those groups
        df_mean = df_mean.merge(valid_groups, on=['time', 'fuel_source_primary'])

    fuels = df_mean['fuel_source_primary'].unique()
    fig = go.Figure()

    # Add bar traces (2 per fuel — HW and non-HW)
    for i, fuel in enumerate(fuels):
        fuel_df = df_mean[df_mean['fuel_source_primary'] == fuel]
        for flag, label, color in [(0, 'Non-Heatwave', 'blue'), (1, 'Heatwave', 'red')]:
            sub_df = fuel_df[fuel_df['EHF_flag'] == flag]
            fig.add_trace(go.Bar(
                x=sub_df['time'],
                y=sub_df['min_max'],
                name=f'{fuel} - {label}',
                marker_color=color,
                visible=(i == 0)
            ))

    # Add demand line trace ONCE, always visible
    demand_df = demand_df.copy()
    demand_df['time'] = pd.to_datetime(demand_df['time'])

    # Normalize demand if needed to match scale
    demand_norm = (demand_df['TOTALDEMAND'] - demand_df['TOTALDEMAND'].min()) / (demand_df['TOTALDEMAND'].max() - demand_df['TOTALDEMAND'].min())

    fig.add_trace(go.Scatter(
        x=demand_df['time'],
        y=demand_df['TOTALDEMAND'],
        mode='lines',
        name='Total Demand (normalized)',
        line=dict(color='green', dash='dash'),
        yaxis='y2',
        visible=True
    ))

    # Build dropdown buttons visibility lists
    buttons = []
    n_bar_traces = 2 * len(fuels)
    for i, fuel in enumerate(fuels):
        vis = [False] * n_bar_traces + [True]  # +1 for demand trace always visible
        vis[2 * i] = True    # Non-heatwave trace for this fuel
        vis[2 * i + 1] = True  # Heatwave trace for this fuel

        buttons.append(dict(
            label=fuel,
            method='update',
            args=[{'visible': vis},
                  {'title': f'Mean Per-Unit Generation: {fuel}'}]
        ))

    # Update layout with secondary y-axis
    fig.update_layout(
        title=f'Mean Per-Unit Generation: {fuels[0]}',
        xaxis_title='Time',
        yaxis=dict(
            title='Normalized Mean Generation',
            range=[0, 1]
        ),
        yaxis2=dict(
            title='Total Demand',
            overlaying='y',
            side='right',
            showgrid=False
        ),
        height=600,
        barmode='group',
        updatemenus=[dict(
            buttons=buttons,
            direction='down',
            x=1.05,
            xanchor='left',
            y=1.15,
            yanchor='top'
        )],
        legend=dict(title='Legend')
    )

    fig.show()


barplot_gen_per_unit(nsw_df, demand[demand['REGIONID'] == 'NSW1'], True)
```
