```python
import cv2
import os

# Set the path to the directory containing images
image_dir = '/g/data/ng72/ms5578/ID_HW_BARRA/data/output/EHF_2017_18/'

# Get a sorted list of image files
image_files = sorted([
    f for f in os.listdir(image_dir)
    if f.lower().endswith(('.png', '.jpg', '.jpeg'))
])

# Check if there are any images
if not image_files:
    raise ValueError("No images found in the specified directory.")

# Read the first image to get frame dimensions
first_image_path = os.path.join(image_dir, image_files[0])
first_image = cv2.imread(first_image_path)
height, width, layers = first_image.shape
frame_size = (width, height)

# Output video path in the same folder
output_video_path = os.path.join(image_dir, 'HW_2017_2018_video.mp4')

# Initialize video writer
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
fps = 2  # Frames per second
video = cv2.VideoWriter(output_video_path, fourcc, fps, frame_size)

# Write each image to the video
for img_file in image_files:
    img_path = os.path.join(image_dir, img_file)
    image = cv2.imread(img_path)
    if image.shape[:2] != (height, width):
        image = cv2.resize(image, frame_size)
    video.write(image)

video.release()
print(f"Video saved to: {output_video_path}")
```
