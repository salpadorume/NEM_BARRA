```python
def jitter(group):
    duplicates = group.groupby(['lat', 'lon']).cumcount()
    
    # Jitter function: add a small random offset
    np.random.seed(42)  # For reproducibility
    jitter_strength = 0.05 # Adjust as needed; degrees latitude/longitude
    
    group['lat_jittered'] = group['lat'] + np.random.uniform(-jitter_strength, jitter_strength, size=len(group)) * duplicates
    group['lon_jittered'] = group['lon'] + np.random.uniform(-jitter_strength, jitter_strength, size=len(group)) * duplicates

    return group
```

```python
def process_group_daily(grp, gen_fpath, hw_tseries, start_date=None, end_date=None):
    """
    Processes a single group of generator data.

    Parameters:
        grp (pd.DataFrame): A single group from gen_details (e.g., from groupby('region')).
        gen_fpath (str): Path to directory containing CSV files named by DUID (e.g., DUID.csv).
        hw_tseries (pd.DataFrame): DataFrame with columns 'time', 'DUID', and data to merge.
        start_date (str): Start date for subsetting the time series.
        end_date (str): End date for subsetting the time series.

    Returns:
        pd.DataFrame: The merged result for the group.
    """
    safe_duids = [duid.replace("/", "_").replace("\\", "_") for duid in grp['DUID']]
    gen_locs = [f"{gen_fpath}/{duid}.csv" for duid in safe_duids]
    dfs = [pd.read_csv(fp,dtype='object') for fp in gen_locs if os.path.exists(fp)]
    print(f"Loaded {len(dfs)} CSV file(s) out of {len(gen_locs)} expected.")

    if not dfs:
        return None

    # This is in case of accidental mid-file headers
    dfs = pd.concat(dfs, ignore_index=True)
    header_row = dfs.columns.tolist()
    dfs = dfs[~dfs.apply(lambda row: list(row) == header_row, axis=1)]

    # This is to correct the column types after removing header rows
    dfs['time'] = pd.to_datetime(dfs['time'])
    dfs['TOTALCLEARED'] = dfs['TOTALCLEARED'].astype(float)
    dfs['TOTALMWh'] = dfs['TOTALMWh'].astype(float)
    dfs['AGCSTATUS'] = pd.to_numeric(dfs['AGCSTATUS'], errors='coerce').fillna(0).astype(int)
    
    dfs['time'] = pd.to_datetime(dfs['time'])
    dfs = dfs.set_index('time').sort_index()
    dfs = dfs.loc[start_date:end_date]

    # Group by DUID and hourly time, sum TOTALMWh
    agg_func = {'TOTALMWh':'sum','TOTALCLEARED':'sum','AGCSTATUS':'max'}
    grouped = dfs.groupby(['DUID', pd.Grouper(freq='1d')]).agg(agg_func).reset_index()

    hw_tseries['time'] = pd.to_datetime(hw_tseries['time'])
    hw_tseries = hw_tseries.set_index(['time']).sort_index()
    hw_tseries = hw_tseries.loc[sdate:edate]
    hw_tseries = hw_tseries.reset_index().set_index(['DUID','time']).sort_index()

    merged = pd.merge_asof(
        grouped.sort_values(by=['time', 'DUID']),
        hw_tseries.sort_values(by=['time', 'DUID']),
        by='DUID',
        on='time',
        tolerance=pd.Timedelta("1d"),
        direction='nearest'
    )

    merged = merged.dropna(how='all')

    return merged
```

```python
def select_group(gen_details, state=None, ftype=None):
    if state is not None and ftype is not None:
        if isinstance(ftype, list):
            grp = gen_details[(gen_details['region'] == state) & (gen_details['fuel_source_primary'].isin(ftype))]
        else:
            groups = gen_details.groupby(['region', 'fuel_source_primary'])
            grp = groups.get_group((state, ftype))
    elif state is not None:
        groups = gen_details.groupby('region')
        grp = groups.get_group(state)
    elif ftype is not None:
        if isinstance(ftype, list):
            grp = gen_details[gen_details['fuel_source_primary'].isin(ftype)]
        else:
            groups = gen_details.groupby('fuel_source_primary')
            grp = groups.get_group(ftype)
    else:
        grp = gen_details

    return grp
```

```python
def clean_df(df, gen_details): 
    df = df.merge(gen_details[['DUID', 'reg_cap_mw','technology_type_primary','fuel_source_primary','lat_jittered','lon_jittered', 'region']], on='DUID', how='left')
    df = df[~((df['TOTALMWh'] < 0))]
    mask = (df['fuel_source_primary'].isin([
            'Water', 'Natural Gas Pipeline', 'Black Coal', 'Coal Seam Methane',
            'Brown Coal', 'Diesel', 'Kerosene'
        ]) & (df['TOTALCLEARED'] <= 0))
    df.loc[mask, 'TOTALMWh'] = np.nan
    
    djf = [12, 1, 2]  # December, January, February
    df = df[df['time'].dt.month.isin(djf)]
 
    return df
```

```python
def min_heatwave_days(df, min_days=10):
    """
    Filters the DataFrame to include only DUIDs with at least `min_days`
    of unique heatwave days (EHF_flag == 1).
    
    Parameters:
        df (pd.DataFrame): Input DataFrame with columns ['DUID', 'time', 'EHF_flag']
        min_days (int): Minimum number of unique heatwave days required

    Returns:
        pd.DataFrame: Filtered DataFrame
    """
    df = df.copy()
    df['time'] = pd.to_datetime(df['time'])
    df['date'] = df['time'].dt.date

    # Count unique heatwave days per DUID
    heatwave_days = (
        df[df['EHF_flag'] == 1]
        .groupby('DUID')['date']
        .nunique()
    )

    # Keep only DUIDs meeting the threshold
    valid_duids = heatwave_days[heatwave_days >= min_days].index
    
    return df[df['DUID'].isin(valid_duids)].copy()
```

```python
def anova_hw(df):
    """
    For each DUID, compare TOTALMWh between EHF_flag==0 and EHF_flag==1 using t-test.
    
    Parameters:
        df (pd.DataFrame): DataFrame with 'DUID', 'TOTALMWh', and 'EHF_flag'.
        
    Returns:
        pd.DataFrame with DUID, t-statistic, p-value, and group means.
    """
    results = []
    
    for duid, group in df.groupby('DUID'):
        # Split by EHF_flag
        group_0 = group[group['EHF_flag'] == 0]['TOTALMWh'].dropna()
        group_1 = group[group['EHF_flag'] == 1]['TOTALMWh'].dropna()
        
        # Skip if one group is empty or too small
        if len(group_0) < 10 or len(group_1) < 10:
            continue
        
        # Perform ANOVA
        t_stat, p_val = stats.f_oneway(group_0, group_1)
        
        results.append({
            'DUID': duid,
            't_statistic': t_stat,
            'p_value': p_val,
            'mean_EHF_0': group_0.mean(),
            'mean_EHF_1': group_1.mean()
        })
    
    return pd.DataFrame(results)
```

```python
def plot_anova_pvalues_map(df, info, anova_results, alpha=0.05,ftypes='Wind and Solar'):
    merged = anova_results.merge(info, on='DUID', how='left').dropna(subset=['lat_jittered', 'lon_jittered'])
    merged['change'] = merged['mean_EHF_1'] - merged['mean_EHF_0']
    merged['size'] = (merged['change'].abs().replace(0, 0.001) * 100) + 20

    def pvalue_significance(p):
        if p < alpha / 10:
            return 'Very strong'
        elif p < alpha:
            return 'Strong'
        elif p < 0.1:
            return 'Weak'
        else:
            return 'Not significant'

    merged['SignificanceLevel'] = merged['p_value'].apply(pvalue_significance)

    def category(row):
        sig = row['SignificanceLevel']
        if sig == 'Not significant':
            return sig
        elif row['change'] < 0:
            return f'{sig} decrease'
        else:
            return f'{sig} increase'

    merged['Category'] = merged.apply(category, axis=1)

    color_map = {
        'Very strong increase': 'darkgreen',
        'Strong increase': 'green',
        'Weak increase': 'lightgreen',
        'Very strong decrease': 'darkred',
        'Strong decrease': 'red',
        'Weak decrease': 'salmon',
        'Not significant': 'lightgray'
    }

    fig = px.scatter_map(
        merged,
        lat='lat_jittered',
        lon='lon_jittered',
        size='size',
        color='Category',
        color_discrete_map=color_map,
        hover_name='DUID',
        hover_data={
            'change': ':.2f',
            'mean_EHF_0': ':.2f',
            'mean_EHF_1': ':.2f',
            'p_value': ':.4f'
        },
        zoom=5,
        map_style='carto-darkmatter',
        title=f'ANOVA Results: Difference in average generation on Heatwave days for {ftypes}'
    )

    fig.update_layout(
        legend_title_text='Significance and Direction',
        margin=dict(l=10, r=10, t=50, b=10)
    )

    fig.show()
```

```python
def bucket_event_days(df, location_col='DUID', event_id_col='event_group', event_day_col='HW_event_day'):
    """
    Assigns each row to a bucket: 'NaN', '1', '2', '3', or 'last'
    where 'last' is the final day of each event for each location.
    
    Requires: event_id_col uniquely identifies events per location.
    """
    df = df.copy()

    # Get last day per event per location
    df['last_day'] = (
        df.groupby([location_col, event_id_col])[event_day_col]
          .transform('max')
    )

    # Define bucket function
    def bucket_day(row):
        if pd.isna(row[event_day_col]):
            return 'NaN'
        elif row[event_day_col] in [1, 2, 3]:
            return str(int(row[event_day_col]))
        elif row[event_day_col] == row['last_day']:
            return 'last'
        else:
            return str(int(row[event_day_col]))  # or could group as 'other'

    df['bucket'] = df.apply(bucket_day, axis=1)

    # Drop helper column if you don’t need it
    df = df.drop(columns=['last_day'])
    return df

# Example usage
data = {
    'location': ['A', 'A', 'A', 'A', 'A', 'A', 'B', 'B', 'B', 'B'],
    'event_id': [1, 1, 1, 2, 2, 2, 1, 1, 2, np.nan],
    'event_day': [1, 2, 3, 1, 2, 4, 1, 2, 1, np.nan]
}

df_bucketed = bucket_event_days(df)
```

```python
def scatter_bucket_vs_nan(df, duid_col, bucket_col, value_col, bucket_to_compare, color_col=None):
    nan_means = (
        df[df[bucket_col] == 'NaN']
        .groupby(duid_col)[value_col]
        .mean()
        .rename('mean_nan')
    )
    bucket_means = (
        df[df[bucket_col] == str(bucket_to_compare)]
        .groupby(duid_col)[value_col]
        .mean()
        .rename(f'mean_{bucket_to_compare}')
    )
    plot_df = pd.concat([nan_means, bucket_means], axis=1).dropna().reset_index()

    if color_col:
        first_vals = df.groupby(duid_col)[color_col].first()
        plot_df = plot_df.merge(first_vals, on=duid_col, how='left')

    fig = px.scatter(
        plot_df,
        x='mean_nan',
        y=f'mean_{bucket_to_compare}',
        color=color_col if color_col else None,
        hover_data=[duid_col] + ([color_col] if color_col else []),
        title=f"{value_col}: NaN days vs. bucket {bucket_to_compare}",
        labels={
            'mean_nan': f'Mean {value_col} on NaN days',
            f'mean_{bucket_to_compare}': f'Mean {value_col} on bucket {bucket_to_compare}'
        },
        trendline="ols"  # Ordinary Least Squares regression line
    )

    # Set equal axes ranges, clip min at zero to avoid negative axis
    min_val = max(0, min(plot_df['mean_nan'].min(), plot_df[f'mean_{bucket_to_compare}'].min()))
    max_val = max(plot_df['mean_nan'].max(), plot_df[f'mean_{bucket_to_compare}'].max())+100

    fig.update_xaxes(range=[min_val, max_val], scaleanchor="y", scaleratio=2)
    fig.update_yaxes(range=[min_val, max_val])

    fig.update_traces(marker=dict(size=10, line=dict(width=1, color='DarkSlateGrey')))
    fig.show()


# # Example usage:

day_1 = scatter_bucket_vs_nan(df_bucketed, 'DUID', 'bucket', 'TOTALMWh', bucket_to_compare='1',color_col='fuel_source_primary')
# day_2 = scatter_bucket_vs_nan(df_bucketed, 'DUID', 'bucket', 'TOTALMWh', bucket_to_compare='2',color_col='fuel_source_primary')
# day_3 = scatter_bucket_vs_nan(df_bucketed, 'DUID', 'bucket', 'TOTALMWh', bucket_to_compare='3',color_col='fuel_source_primary')
# day_n = scatter_bucket_vs_nan(df_bucketed, 'DUID', 'bucket', 'TOTALMWh', bucket_to_compare='last',color_col='fuel_source_primary')
```

```python
def scatter_buckets_with_trendlines(df, duid_col='DUID', event_day_col='event_day',
                                    value_col='TOTALMWh', category_col=None, category_value=None):
    """
    Scatter plot of mean values for buckets 1,2,3,last vs NaN days with trendlines.
    """
    df = df.copy()

    # Optional category filtering
    if category_col is not None and category_value is not None:
        df = df[df[category_col] == category_value]

    # Create bucket column as string
    df['bucket'] = df[event_day_col].apply(lambda x: 'NaN' if pd.isna(x) else str(int(x)))

    # Mark only last day (per DUID) as 'last', keep 1,2,3 as-is
    last_days = df.groupby(duid_col)[event_day_col].transform('max')
    df.loc[df[event_day_col] == last_days, 'bucket'] = 'last'

    # Keep only 1,2,3,last,NaN
    df = df[df['bucket'].isin(['1','2','3','last','NaN'])]

    # Compute mean per DUID per bucket
    nan_means = df[df['bucket'] == 'NaN'].groupby(duid_col)[value_col].mean().rename('mean_nan')
    bucket_means = df[df['bucket'].isin(['1','2','3','last'])].groupby([duid_col, 'bucket'])[value_col].mean().reset_index()
    
    # Merge with NaN means
    plot_df = bucket_means.merge(nan_means, on=duid_col, how='left').dropna().sort_values('bucket')

    # Scatter plot with trendlines
    fig = px.scatter(
        plot_df,
        x='mean_nan',
        y=value_col,
        color='bucket',
        trendline="ols",
        hover_data=[duid_col],
        labels={'mean_nan': f"Mean {value_col} on NaN days"},
        title=f"{value_col}: NaN vs. buckets 1,2,3,last"
    )

    # Set equal axes
    min_val = max(0, plot_df[['mean_nan', value_col]].min().min())
    max_val = plot_df[['mean_nan', value_col]].max().max()+100

    fig.update_yaxes(range=[min_val, max_val])

    fig.show()

scatter_buckets_with_trendlines(
    df=df,
    duid_col='DUID',
    event_day_col='HW_event_day',
    value_col='TOTALMWh',
    category_col='fuel_source_primary',
    category_value='Solar'  # optional filter
)
```

```python
def boxplot_buckets(df, value_col='TOTALMWh', event_day_col='HW_event_day',
                    category_col=None, category_value=None):
    df_plot = df.copy()

    # Optional category filtering
    if category_col and category_value:
        df_plot = df_plot[df_plot[category_col] == category_value]

    # Create buckets: 1,2,3,last
    df_plot['bucket'] = df_plot[event_day_col].apply(
        lambda x: 'NaN' if pd.isna(x) else ('last' if x > 3 else str(int(x)))
    )

    # Keep only relevant buckets
    df_plot = df_plot[df_plot['bucket'].isin(['1','2','3','last'])]

    # Create box plot
    fig = px.box(
        df_plot,
        x='bucket',
        y=value_col,
        points='all',
        color='bucket',
        hover_data=['DUID'],
        title=f"{value_col} distribution per event bucket" + (f" for {category_value}" if category_value else "")
    )
    fig.show()

boxplot_buckets(df, category_col='fuel_source_primary', value_col='norm_std', category_value='Wind')
```

```python
def relative_change_heatmap(df, duid_col, bucket_col, value_col):
    import numpy as np
    nan_means = df[df[bucket_col] == 'NaN'].groupby(duid_col)[value_col].mean().rename('mean_nan')
    mean_df = df.groupby([duid_col, bucket_col])[value_col].mean().reset_index()
    mean_df = mean_df.merge(nan_means, on=duid_col, how='left')
    mean_df['rel_change'] = (mean_df[value_col] - mean_df['mean_nan']) / mean_df['mean_nan']
    plot_df = mean_df[mean_df[bucket_col] != 'NaN'].copy()

    # Create proper bucket order
    buckets = plot_df[bucket_col].unique()
    numeric_buckets = sorted([b for b in buckets if str(b).isdigit()], key=int)
    if 'last' in buckets:
        numeric_buckets.append('last')

    # Convert bucket_col in plot_df to categorical with that order
    plot_df[bucket_col] = pd.Categorical(plot_df[bucket_col], categories=numeric_buckets, ordered=True)

    # Pivot with ordered columns
    heatmap_data = plot_df.pivot(index=duid_col, columns=bucket_col, values='rel_change')

    # Reindex columns to ensure order in heatmap_data
    heatmap_data = heatmap_data.reindex(columns=numeric_buckets)

    fig = px.imshow(
        heatmap_data,
        color_continuous_scale='RdBu',
        origin='lower',
        title=f"Relative change vs Baseline days",
        labels={'color': 'Relative change'},
        aspect='auto'
    )

    # fig.update_layout(height=30*len(plot_df), width=900)
    fig.show()   

for category, group_df in df_bucketed.groupby(['fuel_source_primary','region']):
    print(f"Plotting for category: {category}")
    relative_change_heatmap(group_df, 'DUID', 'bucket', 'TOTALMWh')
```
