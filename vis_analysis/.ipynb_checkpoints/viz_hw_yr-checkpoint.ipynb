```python
def filter_2d_objs(data):
    structure = np.ones((3,3),dtype=int)
    lon, lat, time = data.shape
    filtered = np.zeros_like(data)

    for t in range(time):
        # Label connected components in 2D slice
        labeled, num_features = label(data[:, :, t], structure=structure)
        slices = find_objects(labeled)

        for i, slc in enumerate(slices):
            if slc is None:
                continue
            component_mask = (labeled[slc] == (i + 1))
            size = np.sum(component_mask)
            if size >= 10:
                # Keep the component
                filtered_slice = filtered[:, :, t]
                filtered_slice[slc][component_mask] = data[:, :, t][slc][component_mask]

    return filtered
```

```python
def plot_year_with_shapefile(array, nmap, cmap, norm, gdf):
    lat_min, lat_max = -45, -10
    lon_min, lon_max = 110, 155

    masked = np.ma.masked_where(array == 0, array)

    lons = array.lon.values
    lats = array.lat.values
    lon2d, lat2d = np.meshgrid(lons, lats)

    fig, ax = plt.subplots(figsize=(8, 6),
                           subplot_kw={'projection': ccrs.PlateCarree(central_longitude=180)})

    ax.set_extent([lon_min, lon_max, lat_min, lat_max], crs=ccrs.PlateCarree())

    ax.coastlines()
    ax.add_feature(cartopy.feature.BORDERS)

    # Plot heatwave data
    mesh = ax.pcolormesh(lon2d,
                         lat2d,
                         masked,
                         cmap=cmap,
                         norm=norm,
                         shading='auto',
                         transform=ccrs.PlateCarree())

    # Plot stations
    ax.scatter(nmap['Lon'],
               nmap['Lat'],
               color='black',
               s=25,
               marker="x",
               alpha=0.4,
               transform=ccrs.PlateCarree())

    # Plot shapefile boundaries
    gdf_proj = gdf.to_crs(ccrs.PlateCarree().proj4_init)  # reproject shapefile to PlateCarree
    gdf_proj.boundary.plot(ax=ax, edgecolor='red', linewidth=1, alpha=0.7, zorder=3)

    time_str = pd.to_datetime(array.time.values).strftime('%Y-%m-%d')
    ax.set_title(f"Heatwave Events on {time_str}")

    plt.colorbar(mesh, ax=ax, orientation='vertical', label='Heatwave Index')

    return fig
```

```python
def plot_event(array, cmap, norm, time, nmap):
    lat_min, lat_max = -45, -10
    lon_min, lon_max = 110, 155

    # Subset at time
    labeled_at_time = array.isel(time=time)
    masked = np.ma.masked_where(labeled_at_time == 0, labeled_at_time)

    # Get lat/lon
    lons = array.lon.values
    lats = array.lat.values
    lon2d, lat2d = np.meshgrid(lons, lats)

    # Plot
    fig, ax = plt.subplots(figsize=(8, 6), subplot_kw={'projection': ccrs.PlateCarree(central_longitude=180)})
    ax.set_extent([lon_min, lon_max, lat_min, lat_max], crs=ccrs.PlateCarree())

    ax.coastlines()
    ax.add_feature(cfeature.BORDERS)

    mesh = ax.pcolormesh(lon2d,
                          lat2d, 
                          masked, 
                          cmap=cmap, 
                          norm=norm, 
                          shading='auto', 
                          transform=ccrs.PlateCarree())
    
    ax.scatter(nmap['Lon'],
               nmap['Lat'],
               color='black',
               s=25,
               marker="x",
               alpha=0.4,
               transform=ccrs.PlateCarree())

    objs = np.unique(array)

    cbar = fig.colorbar(mesh, ax=ax, ticks=objs, label="Heatwave Event ID")
    cbar.ax.set_yticklabels([str(i) for i in objs])

    time_str = pd.to_datetime(array.time.values[time]).strftime('%Y-%m-%d')
    ax.set_title(f"Heatwave Events on {time_str}")

    del masked, mesh, ax, labeled_at_time, lon2d, lat2d
    
    return fig
```

```python
def plot_year_with_shapefile(array, nmap, cmap, norm, gdf=shpfile):
    lat_min, lat_max = -45, -10
    lon_min, lon_max = 110, 155

    masked = np.ma.masked_where(array == 0, array)

    lons = array.lon.values
    lats = array.lat.values
    lon2d, lat2d = np.meshgrid(lons, lats)

    fig, ax = plt.subplots(figsize=(8, 6),
                           subplot_kw={'projection': ccrs.PlateCarree(central_longitude=180)})
    ax.set_extent([lon_min, lon_max, lat_min, lat_max], crs=ccrs.PlateCarree())

    ax.coastlines()
    ax.add_feature(cfeature.BORDERS)

    # Plot heatwave data
    mesh = ax.pcolormesh(lon2d,
                         lat2d,
                         masked,
                         cmap=cmap,
                         norm=norm,
                         shading='auto',
                         transform=ccrs.PlateCarree())

    # Plot stations
    ax.scatter(nmap['Lon'],
               nmap['Lat'],
               color='black',
               s=25,
               marker="x",
               alpha=0.4,
               transform=ccrs.PlateCarree())

    # Plot each geometry safely
    for geom in gdf.geometry:
        if geom is None:
            continue
        elif isinstance(geom, Polygon):
            ax.add_geometries([geom], crs=ccrs.PlateCarree(),
                              edgecolor='black', facecolor='none', linewidth=0.3)
        elif isinstance(geom, MultiPolygon):
            ax.add_geometries(geom.geoms, crs=ccrs.PlateCarree(),
                              edgecolor='black', facecolor='none', linewidth=0.3)

    time_str = pd.to_datetime(array.time.values).strftime('%Y-%m-%d')
    ax.set_title(f"Heatwave Events on {time_str}")

    # plt.colorbar(mesh, ax=ax, orientation='vertical', label='Heatwave Index')

    return fig
```

```python
def plot_year(array, nmap,cmap,norm):
    lat_min, lat_max = -45, -10
    lon_min, lon_max = 110, 155

    masked = np.ma.masked_where(array == 0, array)

    # Get lat/lon
    lons = array.lon.values
    lats = array.lat.values
    lon2d, lat2d = np.meshgrid(lons, lats)

    # Plot
    fig, ax = plt.subplots(figsize=(8, 6), subplot_kw={'projection': ccrs.PlateCarree(central_longitude=180)})
    ax.set_extent([lon_min, lon_max, lat_min, lat_max], crs=ccrs.PlateCarree())

    ax.coastlines()
    ax.add_feature(cfeature.BORDERS)

    mesh = ax.pcolormesh(lon2d,
                          lat2d, 
                          masked, 
                          cmap=cmap,
                          norm=norm,
                          shading='auto', 
                          transform=ccrs.PlateCarree())
    
    ax.scatter(nmap['Lon'],
               nmap['Lat'],
               color='black',
               s=25,
               marker="x",
               alpha=0.4,
               transform=ccrs.PlateCarree())

    time_str = pd.to_datetime(array.time.values).strftime('%Y-%m-%d')
    ax.set_title(f"Heatwave Events on {time_str}")

    del masked, mesh, ax, lon2d, lat2d
    
    return fig
```

```python
def plot_ehf(array, nmap, gdf=shpfile):
    lat_min, lat_max = -45, -10
    lon_min, lon_max = 110, 155

    masked = np.ma.masked_where(array == 0, array)

    # Get lat/lon
    lons = array.lon.values
    lats = array.lat.values
    lon2d, lat2d = np.meshgrid(lons, lats)

    # Plot
    fig, ax = plt.subplots(figsize=(8, 6), subplot_kw={'projection': ccrs.PlateCarree(central_longitude=180)})
    ax.set_extent([lon_min, lon_max, lat_min, lat_max], crs=ccrs.PlateCarree())

    ax.coastlines()
    ax.add_feature(cfeature.BORDERS)

    mesh = ax.pcolormesh(lon2d,
                          lat2d, 
                          masked,
                          shading='auto',
                          cmap='magma',
                          transform=ccrs.PlateCarree())
    
    ax.scatter(nmap['Lon'],
               nmap['Lat'],
               color='black',
               s=25,
               marker="x",
               alpha=0.4,
               transform=ccrs.PlateCarree())

    # Plot shapefile boundaries
    gdf_proj = gdf.to_crs(ccrs.PlateCarree().proj4_init)  # reproject shapefile to PlateCarree
    gdf_proj.boundary.plot(ax=ax, edgecolor='red', linewidth=1, alpha=0.7, zorder=3)    

    time_str = pd.to_datetime(array.time.values).strftime('%Y-%m-%d')
    ax.set_title(f"Heatwave Events on {time_str}")

    

    del masked, mesh, ax, lon2d, lat2d
    
    return fig
```
