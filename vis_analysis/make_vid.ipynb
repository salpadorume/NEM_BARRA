```python
import cv2
import os

def crop_center(image, target_width, target_height):
    h, w = image.shape[:2]

    # If image is smaller than target, resize proportionally first
    if w < target_width or h < target_height:
        scale = max(target_width / w, target_height / h)
        image = cv2.resize(image, (int(w * scale), int(h * scale)))
        h, w = image.shape[:2]

    # Crop center
    x_start = (w - target_width) // 2
    y_start = (h - target_height) // 2
    cropped = image[y_start:y_start + target_height, x_start:x_start + target_width]

    return cropped


def make_video_from_images(image_dir, fps=1.7):
    image_files = sorted([
        f for f in os.listdir(image_dir)
        if f.lower().endswith(('.png', '.jpg', '.jpeg'))
    ])

    if not image_files:
        return False

    # First image → frame size
    first_image_path = os.path.join(image_dir, image_files[0])
    first_image = cv2.imread(first_image_path)
    if first_image is None:
        return False
    height, width, _ = first_image.shape
    frame_size = (width, height)

    output_video_path = os.path.join(image_dir, 'video.mp4')
    fourcc = cv2.VideoWriter_fourcc(*'avc1')
    video = cv2.VideoWriter(output_video_path, fourcc, fps, frame_size, True)

    for img_file in image_files:
        img_path = os.path.join(image_dir, img_file)
        image = cv2.imread(img_path)
        if image is None:
            continue
        image = crop_center(image, width, height)  # crop instead of resize
        video.write(image)

    video.release()
    print(f"✅ Video saved to: {output_video_path}")
    return True


def process_all_subdirs(root_dir, fps=1.7):
    for dirpath, _, filenames in os.walk(root_dir):
        if any(f.lower().endswith(('.png', '.jpg', '.jpeg')) for f in filenames):
            print(f"\nProcessing directory: {dirpath}")
            make_video_from_images(dirpath, fps=fps)


if __name__ == "__main__":
    root_directory = '/g/data/ng72/ms5578/ID_HW_BARRA/data/output/alpine_wind/hw/C2'
    process_all_subdirs(root_directory, fps=1.5)
```
